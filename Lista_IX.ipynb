## Exercício 1 — Geração de combinações booleanas

```
def gerar_combinacoes():
    total = 2 ** 3

    for i in range(total):
        binario = format(i, f'0{3}b')
        print(binario)


gerar_combinacoes()
```

### Complexidade

- Complexidade de tempo: **O(n · 2^n)**
- Complexidade de espaço: **O(n)**


## Exercício 2 — Busca Sequencial e Busca Binária

```
def busca_sequencial(lista, chave):
    posicoes = []

    for i in range(len(lista)):
        if lista[i] == chave:
            posicoes.append(i)

    return posicoes


lista = [1, 2, 3, 2, 4, 2, 5]

print(busca_sequencial(lista, 2))
print(busca_sequencial(lista, 8))
```

```
def busca_binaria_multiplos(lista, chave):
    esquerda = 0
    direita = len(lista) - 1
    resultado = []

    while esquerda <= direita:
        meio = (esquerda + direita) // 2

        if lista[meio] == chave:
            i = meio

            while i >= 0 and lista[i] == chave:
                i -= 1

            i += 1

            while i < len(lista) and lista[i] == chave:
                resultado.append(i)
                i += 1

            return resultado

        elif lista[meio] < chave:
            esquerda = meio + 1

        else:
            direita = meio - 1

    return resultado


lista = [1, 2, 2, 2, 3, 4, 5]

print(busca_binaria_multiplos(lista, 2))
print(busca_binaria_multiplos(lista, 8))
```

### Complexidade

Busca sequencial:
- O(n)

Busca binária:
- O(log n + k)

No pior caso:
- O(n)


## Exercício 3 — Fibonacci

```
def fibonacci_iterativo(n):
    if n <= 1:
        return n

    a, b = 0, 1

    for _ in range(2, n + 1):
        a, b = b, a + b

    return b
```

```
def fibonacci_recursivo(n):
    if n <= 1:
        return n

    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)
```

```
memo = {}

def fibonacci_memoizacao(n):
    if n in memo:
        return memo[n]

    if n <= 1:
        return n

    memo[n] = (
        fibonacci_memoizacao(n - 1)
        + fibonacci_memoizacao(n - 2)
    )

    return memo[n]
```

### Complexidades

- Iterativo: O(n)
- Recursivo sem memoização: O(2^n)
- Recursivo com memoização: O(n)


```
import time
import matplotlib.pyplot as plt


def fibonacci_iterativo(n):
    if n <= 1:
        return n

    a, b = 0, 1

    for _ in range(2, n + 1):
        a, b = b, a + b

    return b


def fibonacci_recursivo(n):
    if n <= 1:
        return n

    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)


memo = {}

def fibonacci_memoizacao(n):
    if n in memo:
        return memo[n]

    if n <= 1:
        return n

    memo[n] = (
        fibonacci_memoizacao(n - 1)
        + fibonacci_memoizacao(n - 2)
    )

    return memo[n]


valores_n = [5, 10, 15, 20, 25, 30]

tempos_iterativo = []
tempos_recursivo = []
tempos_memoizacao = []


for n in valores_n:

    inicio = time.time()
    fibonacci_iterativo(n)
    tempos_iterativo.append(time.time() - inicio)

    inicio = time.time()
    fibonacci_recursivo(n)
    tempos_recursivo.append(time.time() - inicio)

    inicio = time.time()
    fibonacci_memoizacao(n)
    tempos_memoizacao.append(time.time() - inicio)


plt.plot(valores_n, tempos_iterativo, label="Iterativo")
plt.plot(valores_n, tempos_recursivo, label="Recursivo")
plt.plot(valores_n, tempos_memoizacao, label="Memoização")

plt.xlabel("Valor de n")
plt.ylabel("Tempo de execução (s)")
plt.title("Comparação de desempenho - Fibonacci")

plt.legend()
plt.grid(True)

plt.show()
```
